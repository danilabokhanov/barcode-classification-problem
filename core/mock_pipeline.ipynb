```python
pd_frames_target_symbology = {}

def build_arggregated_result(
    classifier_key: str,
    classifier_name: str,
    barcodes: List[Barcode],
    target_symbology_frame,
    *, show_summary: bool =True):
    df = pd.DataFrame([{
        "classifier_key": barcode.stats.classifier_keys[classifier_key],
        "classifier_time": barcode.stats.classifier_time,
        "solver_time": barcode.stats.solver_time,
        "barcodes_solved": barcode.is_solved()
    } for barcode in barcodes])
    summary_by_key = df.groupby("classifier_key").agg(
        barcodes_count=("classifier_time", "size"),
        barcodes_solved=("barcodes_solved", "sum"),
        classifier_time_mean=("classifier_time", "mean"),
        classifier_time_median=("classifier_time", "median"),
        solver_time_mean=("solver_time", "mean"),
        solver_time_median=("solver_time", "median")
    ).assign(
        accuracy_percent=lambda x: (
            100 * x["barcodes_solved"] / x["barcodes_count"]
        ).round(3),
        accuracy_percent_median=lambda x: x["accuracy_percent"].median(),
        solver_time_median_median=lambda x: x["solver_time_median"].median()
    ).reset_index()

    time_cols = [
        "classifier_time_mean",
        "classifier_time_median",
        "solver_time_mean",
        "solver_time_median",
        "solver_time_median_median"
    ]
    summary_by_key[time_cols] =  (summary_by_key[time_cols] * 1000).round(3)

    print("Classifier:", classifier_name, "Key:", classifier_key)
    summary_by_key = summary_by_key.sort_values("classifier_key", ascending=True).reset_index(drop=True)

    if classifier_key == TRUE_KEY:
        if show_summary:
            pass
        target_symbology_frame.append(summary_by_key)

def get_saved_barcodes(classifier):
    calculated_res_path = BarcodeStorage.get_filepath_by_pipeline(classifier)
    use_cached_res = classifier.get_use_cached_res()
    barcodes_saved = []
    if use_cached_res:
        barcodes_saved, use_cached_res = BarcodeStorage.load(calculated_res_path)
    print("use_cached_res:", use_cached_res)
    return barcodes_saved, use_cached_res

def run_solution_batched(
    solver: AbstractBarcodeSolver,
    classifiers: List[AbstractBarcodeClassifier],
    /, *, barcodes_pred: List[List[Barcode]],
    barcodes_fit: Optional[List[List[Barcode]]] = None,
    force_overwrite=False
):
    assert len(barcodes_pred) > 0
    assert len(classifiers) == len(barcodes_pred)
    if barcodes_fit is not None:
        assert len(classifiers) == len(barcodes_fit)
    is_cached = []
    for ind in range(len(classifiers)):
        if force_overwrite:
            is_cached.append(False)
            continue
        barcodes_saved, use_cached_res = get_saved_barcodes(classifiers[ind])
        is_cached.append(use_cached_res)
        if use_cached_res:
            barcodes_pred[ind] = barcodes_saved
        else:
            if barcodes_fit is not None and len(barcodes_fit[ind]) > 0:
                for ind in range(len(classifiers)):
                    classifiers[ind].fit(barcodes_fit[ind])

    warmup_cache_classiffier = PerfectClassifier("perfect", "warmup_caches", MARGIN_RATE, use_cached_res=False, batched_mode=True)
    for i in tqdm(range(len(barcodes_pred[0])), desc="Batch progress"):
        is_warmup_done = False
        for classifier_id in range(len(classifiers)):
            if is_cached[classifier_id]:
                continue
            if not is_warmup_done:
                for warmup_iter in range(2):
                    barcode_cp = [deepcopy(barcodes_pred[-1][i])]
                    warmup_cache_classiffier.predict(barcode_cp)
                    solver.solve(barcode_cp)
                is_warmup_done = True
            b = [barcodes_pred[classifier_id][i]]
            classifiers[classifier_id].predict(b)
            solver.solve(b)
            barcodes_pred[classifier_id][i] = b[0]
    for ind in range(len(classifiers)):
        calculated_res_path = BarcodeStorage.get_filepath_by_pipeline(classifiers[ind])
        BarcodeStorage.save(barcodes_pred[ind], calculated_res_path)
    pd_frames_target_symbology[solver.get_name()] = []
    show_summary = solver.get_name() != "ZxingCppLight"
    for ind, classifier in enumerate(classifiers):
        for classifier_key in barcodes_pred[ind][0].stats.classifier_keys:
            build_arggregated_result(
                classifier_key,
                classifier.get_name(),
                barcodes_pred[ind],
                pd_frames_target_symbology[solver.get_name()],
                show_summary=show_summary
            )

```

```python
import os

zxing_cpp_solver: AbstractBarcodeSolver = ZXingCppSolver(MARGIN_RATE, batched_mode=True)
open_cv_solver: AbstractBarcodeSolver = OpenCVSolver(MARGIN_RATE, batched_mode=True)
zbar_solver: AbstractBarcodeSolver = ZBarSolver(MARGIN_RATE, batched_mode=True)
zxing_java_solver: AbstractBarcodeSolver = ZXingJavaSolver(MARGIN_RATE, batched_mode=True)
zxing_cpp_light_solver: AbstractBarcodeSolver = ZXingCppLightSolver(MARGIN_RATE, batched_mode=True)
solvers = [zxing_cpp_solver, open_cv_solver, zbar_solver, zxing_java_solver, zxing_cpp_light_solver]

barcodes, barcodes_1d_qr, barcodes_ean_upc_a_qr = parse_all_datasets()
print("unmatched types:", unmatched_types)
print("barcodes count (train/val/test):", len(barcodes.train), len(barcodes.val), len(barcodes.test))
print("barcodes 1d + qr count (train/val/test):", len(barcodes_1d_qr.train), len(barcodes_1d_qr.val), len(barcodes_1d_qr.test))
print("barcodes ean + upc_a + qr count (train/val/test):", len(barcodes_ean_upc_a_qr.train), len(barcodes_ean_upc_a_qr.val), len(barcodes_ean_upc_a_qr.test))


def plot_dataset_distribution(barcodes_list: List[Barcode], title: str, save_name: str):
    type_counts = {}
    for barcode in barcodes_list:
        tp = barcode.marked_up_type
        type_counts[tp] = type_counts.get(tp, 0) + 1

    sorted_items = sorted(type_counts.items(), key=lambda kv: PRETTY_PRINT_MAP[kv[0]])
    labels = [PRETTY_PRINT_MAP[symbology] for symbology, _ in sorted_items]
    values = [count for _, count in sorted_items]

    plt.figure(figsize=(9, 5))
    plt.bar(labels, values)
    plt.ylabel("Barcodes count", fontsize=11)
    plt.gca().yaxis.set_major_locator(MaxNLocator(nbins=12))
    plt.xlabel("Symbology", fontsize=11)
    plt.title(title)
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    os.makedirs(GRAPHS_DIR, exist_ok=True)
    plt.savefig(f"{GRAPHS_DIR}/{save_name}", format="png", pad_inches=0)
    plt.show()
    plt.close()


all_barcodes = barcodes.train + barcodes.val + barcodes.test
plot_dataset_distribution(all_barcodes, "Dataset symbology distribution (full)", "symbology_distribution_full.png")
plot_dataset_distribution(barcodes.test, "Dataset symbology distribution (test)", "symbology_distribution_test.png")
```

```python
datasets = [barcodes, barcodes_ean_upc_a_qr, barcodes_1d_qr, barcodes, barcodes]
assert len(solvers) == len(datasets)
rnd_adds_ranges = [MAX_ADDITIONAL_TYPES, MAX_ADDITIONAL_TYPES_OPEN_CV, MAX_ADDITIONAL_TYPES_ZBAR, MAX_ADDITIONAL_TYPES, MAX_ADDITIONAL_TYPES]
assert len(solvers) == len(rnd_adds_ranges)
barcodes_for_solvers = []
mock_basic_classifiers = []
subset_decompositions = {
    "2_11": get_all_partitions_2_11_decomposition(),
    "2_43": get_all_partitions_2_43_decomposition(),
    "2_46": get_all_partitions_2_46_decomposition(),
    "2_67": get_all_partitions_2_67_decomposition(),
    "2_75": get_all_partitions_2_75_decomposition(),
    "2_99": get_all_partitions_2_99_decomposition(),
    "2_100": get_all_partitions_2_100_decomposition(),
    "2_101": get_all_partitions_2_101_decomposition(),
    "2_107": get_all_partitions_2_107_decomposition(),
    "2_108": get_all_partitions_2_108_decomposition(),
    "2_109": get_all_partitions_2_109_decomposition(),
    "2_110": get_all_partitions_2_110_decomposition(),
    "2_115": get_all_partitions_2_115_decomposition(),
    "2_116": get_all_partitions_2_116_decomposition(),
    "2_123": get_all_partitions_2_123_decomposition(),
    "2_124": get_all_partitions_2_124_decomposition(),
    "3_579": get_all_partitions_3_579_decomposition(),
    "3_621": get_all_partitions_3_621_decomposition(),
    "3_645": get_all_partitions_3_645_decomposition(),
    "3_648": get_all_partitions_3_648_decomposition(),
    "3_654": get_all_partitions_3_654_decomposition(),
    "3_672": get_all_partitions_3_672_decomposition(),
    "3_681": get_all_partitions_3_681_decomposition(),
    "3_690": get_all_partitions_3_690_decomposition(),
    "3_699": get_all_partitions_3_699_decomposition(),
    "3_708": get_all_partitions_3_708_decomposition(),
    "3_810": get_all_partitions_3_810_decomposition(),
    "3_837": get_all_partitions_3_837_decomposition(),
    "3_864": get_all_partitions_3_864_decomposition(),
    "3_873": get_all_partitions_3_873_decomposition(),
    "3_882": get_all_partitions_3_882_decomposition(),
    "3_888": get_all_partitions_3_888_decomposition(),
    "3_891": get_all_partitions_3_891_decomposition(),
    "3_942": get_all_partitions_3_942_decomposition(),
    "4_642": get_all_partitions_4_642_decomposition(),
    "4_643": get_all_partitions_4_643_decomposition(),
    "4_659": get_all_partitions_4_659_decomposition(),
    "4_671": get_all_partitions_4_671_decomposition(),
    "4_673": get_all_partitions_4_673_decomposition(),
    "4_751": get_all_partitions_4_751_decomposition(),
    "4_765": get_all_partitions_4_765_decomposition(),
    "4_825": get_all_partitions_4_825_decomposition(),
    "4_839": get_all_partitions_4_839_decomposition(),
    "4_924": get_all_partitions_4_924_decomposition(),
    "4_928": get_all_partitions_4_928_decomposition(),
    "4_929": get_all_partitions_4_929_decomposition(),
    "4_931": get_all_partitions_4_931_decomposition(),
    "4_933": get_all_partitions_4_933_decomposition(),
    "4_987": get_all_partitions_4_987_decomposition(),
    "4_1084": get_all_partitions_4_1084_decomposition(),
    "4_1103": get_all_partitions_4_1103_decomposition(),
    "4_1106": get_all_partitions_4_1106_decomposition(),
    "4_1108": get_all_partitions_4_1108_decomposition(),
    "4_1259": get_all_partitions_4_1259_decomposition(),
    "4_1466": get_all_partitions_4_1466_decomposition(),
    "4_1468": get_all_partitions_4_1468_decomposition(),
    "4_1474": get_all_partitions_4_1474_decomposition(),
    "4_1476": get_all_partitions_4_1476_decomposition(),
    "4_1498": get_all_partitions_4_1498_decomposition(),
    "4_1642": get_all_partitions_4_1642_decomposition(),
    "4_1658": get_all_partitions_4_1658_decomposition(),
    "4_1690": get_all_partitions_4_1690_decomposition(),
    "5_179": get_all_partitions_5_179_decomposition(),
    "5_198": get_all_partitions_5_198_decomposition(),
    "5_263": get_all_partitions_5_263_decomposition(),
    "5_415": get_all_partitions_5_415_decomposition(),
    "5_420": get_all_partitions_5_420_decomposition(),
    "5_460": get_all_partitions_5_460_decomposition(),
    "5_461": get_all_partitions_5_461_decomposition(),
    "5_590": get_all_partitions_5_590_decomposition(),
    "5_594": get_all_partitions_5_594_decomposition(),
    "5_609": get_all_partitions_5_609_decomposition(),
    "5_614": get_all_partitions_5_614_decomposition(),
    "5_618": get_all_partitions_5_618_decomposition(),
    "5_637": get_all_partitions_5_637_decomposition(),
    "5_652": get_all_partitions_5_652_decomposition(),
    "5_672": get_all_partitions_5_672_decomposition(),
    "5_677": get_all_partitions_5_677_decomposition(),
    "5_680": get_all_partitions_5_680_decomposition(),
    "5_731": get_all_partitions_5_731_decomposition(),
    "5_741": get_all_partitions_5_741_decomposition(),
    "5_890": get_all_partitions_5_890_decomposition(),
    "6_9": get_all_partitions_6_9_decomposition(),
    "6_29": get_all_partitions_6_29_decomposition(),
    "6_31": get_all_partitions_6_31_decomposition(),
    "6_44": get_all_partitions_6_44_decomposition(),
    "6_45": get_all_partitions_6_45_decomposition(),
    "6_47": get_all_partitions_6_47_decomposition(),
    "6_48": get_all_partitions_6_48_decomposition(),
    "6_49": get_all_partitions_6_49_decomposition(),
    "6_83": get_all_partitions_6_83_decomposition(),
    "6_84": get_all_partitions_6_84_decomposition(),
    "6_85": get_all_partitions_6_85_decomposition(),
    "6_104": get_all_partitions_6_104_decomposition(),
    "6_105": get_all_partitions_6_105_decomposition(),
    "6_220": get_all_partitions_6_220_decomposition(),
    "6_221": get_all_partitions_6_221_decomposition(),
    "6_234": get_all_partitions_6_234_decomposition(),
    "6_240": get_all_partitions_6_240_decomposition(),
    "6_246": get_all_partitions_6_246_decomposition(),
    "7_2": get_all_partitions_7_2_decomposition(),
    "7_7": get_all_partitions_7_7_decomposition(),
    "7_8": get_all_partitions_7_8_decomposition(),
    "7_9": get_all_partitions_7_9_decomposition(),
    "7_11": get_all_partitions_7_11_decomposition(),
    "7_12": get_all_partitions_7_12_decomposition(),
    "7_13": get_all_partitions_7_13_decomposition(),
    "7_14": get_all_partitions_7_14_decomposition(),
    "7_15": get_all_partitions_7_15_decomposition(),
    "7_16": get_all_partitions_7_16_decomposition(),
    "7_17": get_all_partitions_7_17_decomposition(),
    "7_24": get_all_partitions_7_24_decomposition(),
    "7_25": get_all_partitions_7_25_decomposition(),
    "8_0": get_all_partitions_8_0_decomposition(),
}

for i in range(len(solvers)):
    solver_name = solvers[i].get_name()
    perfect_classifier: AbstractBarcodeClassifier = PerfectClassifier("perfect", solver_name, MARGIN_RATE, use_cached_res=True, batched_mode=True)
    rand_adds_classifiers: List[AbstractBarcodeClassifier] = [RandomAdditionalsClassifier(
        f"random_additionals_{additional_types}", solver_name, MARGIN_RATE, addinional_types_num=additional_types, use_cached_res=True, batched_mode=True)\
        for additional_types in range(rnd_adds_ranges[i], 0, -1)]
    if solver_name != "ZxingCppLight":
        mock_basic_classifiers.append(rand_adds_classifiers + [perfect_classifier])
    else:
        mock_basic_classifiers.append([])
    if solver_name == "ZxingCpp":
        accurate_subset_classifiers: List[AbstractBarcodeClassifier] = [
            AccurateSubsetClassifier(dec_key, solver_name, MARGIN_RATE, dec_val[solver_name], use_cached_res=True, batched_mode=True)
            for dec_key, dec_val in subset_decompositions.items()
        ]
        mock_basic_classifiers[-1].extend(accurate_subset_classifiers)
    elif solver_name == "ZxingCppLight":
        from utils.subsets_decomposition import ALL_TYPES

        all_types = list(ALL_TYPES["ZxingCpp"])
        all_types_set = set(all_types)

        ean13 = BarcodeType.EAN_13
        upc_a = BarcodeType.UPC_A
        ean8 = BarcodeType.EAN_8
        upc_e = BarcodeType.UPC_E

        def is_valid_side(side: set) -> bool:
            if not side or side == all_types_set:
                return False
            return True

        K = ZxingCppLight_NUM_CLASSES

        atomic_units = [
            [ean13, upc_a, ean8, upc_e],
            *[[tp] for tp in all_types if tp not in {ean13, upc_a, ean8, upc_e}]
        ]
        n_units = len(atomic_units)
        if K < 2 or K > n_units:
            raise ValueError(f"ZxingCppLight_NUM_CLASSES must be in 2..{n_units}")

        def gen_set_partitions(items, k):
            items = list(items)
            first = items[0]

            def rec(i, groups):
                m = len(groups)
                remaining = len(items) - i
                if m > k or m + remaining < k:
                    return
                if i == len(items):
                    if m == k:
                        yield [g[:] for g in groups]
                    return

                item = items[i]

                for gi in range(m):
                    groups[gi].append(item)
                    yield from rec(i + 1, groups)
                    groups[gi].pop()

                if m < k:
                    groups.append([item])
                    yield from rec(i + 1, groups)
                    groups.pop()

            groups = [[first]]
            yield from rec(1, groups)

        decompositions = []

        def append_partition_groups(groups):
            subsets = []
            for group in groups:
                flat = []
                for unit in group:
                    flat.extend(unit)
                subsets.append(sorted(flat, key=lambda t: t.name))
            subsets = sorted(subsets, key=lambda subset: tuple(t.name for t in subset))
            decompositions.append(subsets)

        for groups in gen_set_partitions(atomic_units, K):
            append_partition_groups(groups)

        zxing_cpp_light_decompositions = decompositions

        light_subset_classifiers: List[AbstractBarcodeClassifier] = [
            AccurateSubsetClassifier(
                f"{ZxingCppLight_NUM_CLASSES}_{idx}",
                solver_name,
                MARGIN_RATE,
                subsets,
                use_cached_res=True,
                batched_mode=True,
            )
            for idx, subsets in enumerate(decompositions)
        ]
        mock_basic_classifiers[-1].extend(light_subset_classifiers)
    barcodes_for_solvers.append([])
    if solver_name == "ZxingCppLight":
        light_rng = random.Random(42)
        light_test_n = max(1, int(0.5 * len(datasets[i].test)))
        light_test_sample = light_rng.sample(datasets[i].test, light_test_n)
    else:
        light_test_sample = None

    for j in range(len(mock_basic_classifiers[-1])):
        if light_test_sample is not None:
            barcodes_for_solvers[-1].append(deepcopy(light_test_sample))
        else:
            barcodes_for_solvers[-1].append(deepcopy(datasets[i].test))
```

```python
def reorder_index_priority(df: pd.DataFrame, priority: list[str] = [
    PRETTY_PRINT_MAP[BarcodeType.AZTEC], PRETTY_PRINT_MAP[BarcodeType.DATA_MATRIX], PRETTY_PRINT_MAP[BarcodeType.PDF_417], PRETTY_PRINT_MAP[BarcodeType.QR_CODE]
]) -> pd.DataFrame:
    priority_rows = [x for x in priority if x in df.index]
    rest = [x for x in df.index if x not in priority]
    return df.loc[priority_rows + rest]
```

```python
def build_stat(index_map, column):
    df = pd.concat(
        {
            key: dfs[index_map[key]]
                .assign(classifier_key=lambda d: d["classifier_key"].map(PRETTY_PRINT_MAP))
                .set_index("classifier_key")[column]
            for key, dfs in pd_frames_target_symbology.items() if key != "ZxingCppLight"
        },
        axis=1
    )
    df.index.name = "Symbology"
    return reorder_index_priority(df)
```

```python
def build_rand_adds_stat(solver_idx, column):
    classifiers = mock_basic_classifiers[solver_idx]
    solver_name = solvers[solver_idx].get_name()
    series = {}
    for i, clf in enumerate(classifiers):
        if isinstance(clf, RandomAdditionalsClassifier):
            k = clf.get_addinional_types_num()
        elif isinstance(clf, PerfectClassifier):
            k = 0
        else:
            continue
        df = pd_frames_target_symbology[solver_name][i]
        s = (
            df.assign(classifier_key=lambda d: d["classifier_key"].map(PRETTY_PRINT_MAP))
              .set_index("classifier_key")[column]
        )
        series[k] = s
    stat = pd.concat(series, axis=1).sort_index(axis=1, ascending=True)
    stat.index.name = "Symbology"

    return reorder_index_priority(stat)
```

```python
def normalize_by_k0(df: pd.DataFrame) -> pd.DataFrame:
    if 0 not in df.columns:
        raise ValueError("Column k=0 not found in dataframe")

    normalized = df.div(df[0], axis=0).round(3)
    return normalized
```

```python
def build_subset_summary_stat(solver_idx, clf_start_idx):
    solver_name = solvers[solver_idx].get_name()
    classifiers = mock_basic_classifiers[solver_idx]

    results = {}

    for i in range(clf_start_idx, len(classifiers)):
        clf = classifiers[i]
        clf_name = clf.get_name()
        df = pd_frames_target_symbology[solver_name][i]
        df = (
            df.assign(classifier_key=lambda d: d["classifier_key"].map(PRETTY_PRINT_MAP))
              .set_index("classifier_key")
        )
        time_row = df["solver_time_median"]
        acc_row = df["accuracy_percent"]
        stat = pd.DataFrame(
            [time_row, acc_row],
            index=["T", "A"]
        )
        stat = reorder_index_priority(stat.T).T
        stat["Median"] = stat.median(axis=1)
        stat["Mean"] = stat.mean(axis=1)
        results[clf_name] = stat

    return results
```

```python
pd_subset_summary_stat = build_subset_summary_stat(0, MAX_ADDITIONAL_TYPES + 1)

subset_time_summary = pd.DataFrame({
    clf_name: df.loc["T"] for clf_name, df in pd_subset_summary_stat.items()
})
subset_time_summary = subset_time_summary.round(3)
subset_time_summary.index.name = "Symbology"
subset_time_summary.columns.name = "Classifier"

subset_accuracy_summary = pd.DataFrame({
    clf_name: df.loc["A"] for clf_name, df in pd_subset_summary_stat.items()
})
subset_accuracy_summary = subset_accuracy_summary.round(3)
subset_accuracy_summary.index.name = "Symbology"
subset_accuracy_summary.columns.name = "Classifier"

dataframe_to_latex_table(
    subset_time_summary,
    table_name="subset_summary_time_all",
    caption="(ZxingCpp) Временные метрики для всех subset_decompositions",
    label="tab:subset_summary_time_all"
)
print("subset_summary_time_all")
display(subset_time_summary)

dataframe_to_latex_table(
    subset_accuracy_summary,
    table_name="subset_summary_accuracy_all",
    caption="(ZxingCpp) Метрики точности для всех subset_decompositions",
    label="tab:subset_summary_accuracy_all"
)
print("subset_summary_accuracy_all")
display(subset_accuracy_summary)

quality_perfect_median = (
    5 * pd_frames_summary_stat["perfect"].loc["SolverTime (Median)", "ZxingCpp"]
    - 1 * pd_frames_summary_stat["perfect"].loc["Accuracy (Median)", "ZxingCpp"]
)
quality_perfect_mean = (
    3 * pd_frames_summary_stat["perfect"].loc["SolverTime (Mean)", "ZxingCpp"]
    - 1 * pd_frames_summary_stat["perfect"].loc["Accuracy (Mean)", "ZxingCpp"]
)

subset_combined_summary = pd.DataFrame({
    "SolverTime (Median)": subset_time_summary.loc["Median"],
    "SolverTime (Mean)": subset_time_summary.loc["Mean"],
    "Accuracy (Median)": subset_accuracy_summary.loc["Median"],
    "Accuracy (Mean)": subset_accuracy_summary.loc["Mean"],
}).T
subset_combined_summary.loc[r"$\mathcal{X}$ (Median)"] = (
    quality_perfect_median / (
        5 * subset_combined_summary.loc["SolverTime (Median)"]
        - 1 * subset_combined_summary.loc["Accuracy (Median)"]
    )
).round(3)
subset_combined_summary.loc[r"$\mathcal{X}$ (Mean)"] = (
    quality_perfect_mean / (
        3 * subset_combined_summary.loc["SolverTime (Mean)"]
        - 1 * subset_combined_summary.loc["Accuracy (Mean)"]
    )
).round(3)
subset_combined_summary = subset_combined_summary.round(3)
subset_combined_summary.index.name = "Metric"
subset_combined_summary.columns.name = "Classifier"

dataframe_to_latex_table(
    subset_combined_summary,
    table_name="subset_combined_summary_zxingcpp",
    caption="(ZxingCpp) Итоговые метрики качества subset_decompositions",
    label="tab:subset_combined_summary_zxingcpp"
)
print("subset_combined_summary_zxingcpp")
display(subset_combined_summary)

solver_idx_light = next(i for i, s in enumerate(solvers) if s.get_name() == "ZxingCppLight")
light_subset_summary = build_subset_summary_stat(solver_idx_light, 0)

ranked = []
for clf_name, stat in light_subset_summary.items():
    idx = int(clf_name.split("_")[-1])
    median_time = float(stat.loc["T", "Median"])
    mean_time = float(stat.loc["T", "Mean"])
    subsets = zxing_cpp_light_decompositions[idx]
    ranked.append({
        "classifier_name": clf_name,
        "median_of_medians_solver_time_ms": median_time,
        "mean_of_medians_solver_time_ms": mean_time,
        "subsets": " || ".join(",".join(PRETTY_PRINT_MAP[t] for t in subset) for subset in subsets),
    })

pd_light_top10 = (
    pd.DataFrame(ranked)
    .sort_values("median_of_medians_solver_time_ms", ascending=True)
    .head(10)
    .reset_index(drop=True)
)
print("ZxingCppLight top-10 partitions (fastest by median-of-medians):")
with pd.option_context("display.max_colwidth", None, "display.max_columns", None, "display.width", 4000):
    display(pd_light_top10[["classifier_name", "median_of_medians_solver_time_ms", "mean_of_medians_solver_time_ms", "subsets"]])

pd_light_top10_mean = (
    pd.DataFrame(ranked)
    .sort_values("mean_of_medians_solver_time_ms", ascending=True)
    .head(10)
    .reset_index(drop=True)
)
print("ZxingCppLight top-10 partitions (fastest by mean-of-medians):")
with pd.option_context("display.max_colwidth", None, "display.max_columns", None, "display.width", 4000):
    display(pd_light_top10_mean[["classifier_name", "median_of_medians_solver_time_ms", "mean_of_medians_solver_time_ms", "subsets"]])
```

```python
from tools.tex_tables_builder import paired_dataframes_to_latex_table, paired_group_assignment_to_latex_table, save_group

TOP_K = 5
SYMBOLOGY_ROWS = [r for r in subset_time_summary.index if r not in ("Median", "Mean")]
TWO_DIM_COUNT  = 4

MEDIAN_ROWS  = ["SolverTime (Median)", "Accuracy (Median)", r"$\mathcal{X}$ (Median)"]
MEAN_ROWS    = ["SolverTime (Mean)",   "Accuracy (Mean)",   r"$\mathcal{X}$ (Mean)"]
DISPLAY_ROWS = ["T (ms)", "A (%)", r"$\mathcal{X}$"]


def build_decomposition_frame(classifier_names, subset_decompositions, pretty_print_map):
    rows = []
    for clf_name in classifier_names:
        subsets = subset_decompositions[clf_name]["ZxingCpp"]
        decomp_str = " | ".join(
            ", ".join(pretty_print_map[t] for t in subset)
            for subset in subsets
        )
        rows.append({"Classifier": clf_name, "Decomposition": decomp_str})
    df = pd.DataFrame(rows, index=range(1, len(rows) + 1))
    df.index.name = "Rank"
    return df


def build_group_assignment_frame(clf_names, subset_decompositions, pretty_print_map, solver="ZxingCpp"):
    all_types = sorted(
        set(t for clf in clf_names for group in subset_decompositions[clf][solver] for t in group),
        key=lambda t: pretty_print_map[t]
    )
    data = {}
    for clf_name in clf_names:
        groups = subset_decompositions[clf_name][solver]
        type_to_group = {t: i + 1 for i, grp in enumerate(groups) for t in grp}
        data[clf_name] = {pretty_print_map[t]: type_to_group.get(t) for t in all_types}
    df = pd.DataFrame(data, index=[pretty_print_map[t] for t in all_types])
    df.index.name = "Symbology"
    return reorder_index_priority(df)


for n_groups in range(2, 9):
    prefix = f"{n_groups}_"
    cols_n = [c for c in subset_combined_summary.columns if str(c).startswith(prefix)]
    if not cols_n:
        continue

    k = min(TOP_K, len(cols_n))
    sub_quality = subset_combined_summary.loc[[r"$\mathcal{X}$ (Median)", r"$\mathcal{X}$ (Mean)"], cols_n]
    top_median = sub_quality.loc[r"$\mathcal{X}$ (Median)"].sort_values(ascending=True).head(k).index.tolist()
    top_mean   = sub_quality.loc[r"$\mathcal{X}$ (Mean)"].sort_values(ascending=True).head(k).index.tolist()

    print(f"\n{'='*60}")
    print(f"N_groups = {n_groups}")
    print(f"Top-{k} by Quality (Median): {top_median}")
    print(f"Top-{k} by Quality (Mean):   {top_mean}")

    for table_suffix, df_source, caption_suffix, measure_unit in [
        ("time",     subset_time_summary,    "время работы", "ms"),
        ("accuracy", subset_accuracy_summary, "точность", "\%"),
    ]:
        paired_dataframes_to_latex_table(
            df_source.loc[SYMBOLOGY_ROWS, top_median],
            df_source.loc[SYMBOLOGY_ROWS, top_mean],
            left_header=f"Median ({measure_unit})", right_header=f"Mean ({measure_unit})",
            table_name=f"top{k}_{table_suffix}_{n_groups}groups",
            caption=f"(ZxingCpp) Топ-{k} разбиений на {n_groups} групп — {caption_suffix}",
            label=f"tab:top{k}_{table_suffix}_{n_groups}groups",
            section_after=TWO_DIM_COUNT,
            group=RAND_ADDS_GROUP,
        )

    summary_left  = subset_combined_summary.loc[MEDIAN_ROWS, top_median].copy()
    summary_right = subset_combined_summary.loc[MEAN_ROWS,   top_mean].copy()
    summary_left.index  = DISPLAY_ROWS
    summary_right.index = DISPLAY_ROWS
    summary_left.index.name  = "Metric"
    summary_right.index.name = "Metric"
    paired_dataframes_to_latex_table(
        summary_left, summary_right,
        left_header="Median", right_header="Mean",
        table_name=f"top{k}_summary_{n_groups}groups",
        caption=f"(ZxingCpp) Топ-{k} разбиений на {n_groups} групп — сводные метрики",
        label=f"tab:top{k}_summary_{n_groups}groups",
        group=RAND_ADDS_GROUP,
    )

    dec_left  = build_decomposition_frame(top_median, subset_decompositions, PRETTY_PRINT_MAP)
    dec_right = build_decomposition_frame(top_mean,   subset_decompositions, PRETTY_PRINT_MAP)

    assign_left  = build_group_assignment_frame(top_median, subset_decompositions, PRETTY_PRINT_MAP)
    assign_right = build_group_assignment_frame(top_mean,   subset_decompositions, PRETTY_PRINT_MAP)
    paired_group_assignment_to_latex_table(
        assign_left, assign_right,
        left_header="Median", right_header="Mean",
        table_name=f"top{k}_assignment_{n_groups}groups",
        caption=f"(ZxingCpp) Топ-{k} разбиений на {n_groups} групп — принадлежность символик к группам",
        label=f"tab:top{k}_assignment_{n_groups}groups",
        section_after=TWO_DIM_COUNT,
        group=RAND_ADDS_GROUP,
    )

    print(f"Summary (Median ranking):")
    display(summary_left.round(3))
    print(f"Summary (Mean ranking):")
    display(summary_right.round(3))
    print(f"Decompositions (Median ranking):")
    display(dec_left)
    print(f"Decompositions (Mean ranking):")
    display(dec_right)
    print(f"Group assignments (Median ranking):")
    display(assign_left)
    print(f"Group assignments (Mean ranking):")
    display(assign_right)

save_group(RAND_ADDS_GROUP, RAND_ADDS_GROUP)
```
